# Ejecutar el MILP con B sensores

```python
from pathlib import Path

def run_milp_from_notebook(
    B: int = 20,
    selection_dir: Path = DATA_PROCESSED,
    net_path: Path = DATA_RAW / "simulation" / "algeciras.net.xml",
) -> dict:
    """
    Ejecuta el MILP con B sensores, calcula cobertura y genera:

      - selected_edges_B{B}.txt  (para selección en SUMO)
      - sensors_B{B}.add.xml     (POIs de sensores para SUMO)

    Devuelve un dict con:
      - result, df_cov, selection_path, pois_path
    """
    print(f"=== Ejecutando MILP con B = {B} sensores ===")

    result = solve_sensor_placement(
        max_sensors=B,
        msg=True,  # muestra salida del solver en la consola del notebook
    )

    print(f"\nEstado del solver: {result['status']}")
    print(f"Valor objetivo (residuos + slacks): {result['objective_value']}")
    print(f"Número de arcos con sensor: {len(result['selected_edges'])} (B = {B})")

    # Cobertura
    df_cov = compute_flow_coverage(milp_inputs, result["selected_edges"])
    print("\n=== Cobertura de flujo ===")
    display(df_cov)

    # Guardar selección .txt
    selection_dir.mkdir(parents=True, exist_ok=True)
    sel_path = selection_dir / f"selected_edges_B{B}.txt"
    with sel_path.open("w", encoding="utf-8") as f:
        for eid in result["selected_edges"]:
            f.write(f"edge:{eid}\n")
    print(f"\nSelección de edges guardada en: {sel_path}")

    # Generar POIs .add.xml
    pois_path = selection_dir / f"sensors_B{B}.add.xml"
    make_sensor_pois_from_txt(
        selected_txt=sel_path,
        net_xml=net_path,
        output_add=pois_path,
        color="1,1,0",  # amarillo
        size=6.0,       # 3x o más grandes
    )

    return {
        "result": result,
        "df_cov": df_cov,
        "selection_path": sel_path,
        "pois_path": pois_path,
    }

# Ejemplo: lanzar MILP con 20 sensores
milp_run = run_milp_from_notebook(B=30)
```

# Dashboard sencillo de imágenes de configuración de sensores

```python
from IPython.display import display, Image
import ipywidgets as widgets

def show_milp_image(B: int):
    """
    Muestra la imagen de resultados del MILP almacenada en:
        data/Results/{B}_MILP.png
    """
    img_path = RESULTS_DIR / f"{B}_MILP.png"
    if not img_path.exists():
        print(f"No se encontró la imagen {img_path}")
        return
    display(Image(filename=str(img_path)))

# Widget: deslizador para elegir B (ajusta el rango a tus valores disponibles)
B_slider = widgets.IntSlider(
    value=20,
    min=5,
    max=40,
    step=5,
    description="Sensors B:",
    continuous_update=False,
)

ui = widgets.VBox([B_slider])

out = widgets.Output()

def on_B_change(change):
    if change["name"] == "value":
        B = change["new"]
        out.clear_output()
        with out:
            print(f"Configuración de sensores para B = {B}")
            show_milp_image(B)

B_slider.observe(on_B_change, names="value")

display(ui, out)

# Llamada inicial
on_B_change({"name": "value", "new": B_slider.value})
```
